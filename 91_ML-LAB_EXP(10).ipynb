```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Training data
reviews = [
    "The food was delicious",
    "Amazing food and friendly staff",
    "Excellent service and tasty food",
    "The food was cold and tasteless",
    "Terrible service",
    "Poor food and bad experience"
]

labels = [
    "Positive",
    "Positive",
    "Positive",
    "Negative",
    "Negative",
    "Negative"
]

# Convert text into numerical features
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(reviews)

# Train Naive Bayes classifier
model = MultinomialNB()
model.fit(X, labels)

# Test reviews
test_reviews = [
    "The food was tasty and the service was excellent",
    "The food was bad and the service was terrible"
]

X_test = vectorizer.transform(test_reviews)

# Predict sentiment
predictions = model.predict(X_test)

for review, sentiment in zip(test_reviews, predictions):
    print(review, "->", sentiment)
```
